```python
import pandas as pd

customers = pd.DataFrame({
    'CustomerID': [1, 2, 3, 4],
    'CustomerName': ['Alice', 'Bob', 'Charlie', 'David'],
    'City': ['London', 'Manchester', 'Leeds', 'Bristol']
})

products = pd.DataFrame({
    'ProductID': [101, 102, 103],
    'ProductName': ['Laptop', 'Mouse', 'Keyboard'],
    'Price': [900, 25, 60]
})

orders = pd.DataFrame({
    'OrderID': [1, 2, 3, 4],
    'CustomerID': [1, 1, 2, 3],
    'ProductID': [101, 102, 103, 102],
    'Quantity': [1, 2, 1, 3]
})

df = orders.merge(customers, on = 'CustomerID', how = 'left')
df = df.merge(products, on = 'ProductID', how = 'left')

#create sale column
df['Sales'] = df['Quantity'] * df['Price']
df
```

```python
###Product Analysis
product_analysis = (
    df.groupby('ProductName')
      .agg(
          TotalQuantity = ('Quantity', 'sum'),
          TotalSales = ('Sales', 'sum')
      )
      .reset_index()
)

product_analysis.sort_values(
    'TotalSales',
    ascending = False
)
```

```python
###Customer Analysis
customer_analysis = (
    df.groupby('CustomerName')
    .agg(
        TotalOrders = ('OrderID', 'count'),
        TotalQuantity = ('Quantity', 'sum'),
        TotalSales = ('Sales', 'sum')
    )
    .reset_index()
)

customer_analysis.sort_values(
    'TotalSales',
    ascending = False
)
```

```python
###City Analysis
city_analysis = (
    df.groupby('City')
    .agg(
        TotalQuantity = ('Quantity', 'sum'),
        TotalSales = ('Sales', 'sum')
    )
    .reset_index()
)

city_analysis.sort_values(
    'TotalSales',
    ascending = False
)
```

```python
###Revenue Share
customer_analysis['RevenueShare'] = (
    customer_analysis['TotalSales'] / customer_analysis['TotalSales'].sum() * 100
)

customer_analysis.sort_values(
    'RevenueShare',
    ascending = False
)
```
