# Retail Sales Performance Analysis

## Business Objective

This project analyses retail sales performance to identify key revenue contributors, high-value customers, and strong-performing markets.

The analysis aims to provide data-driven insights that can support product, customer, and marketing decisions.

## Tools

- Python
- Pandas
- Jupyter Notebook

## Dataset Overview

The dataset contains three related tables:

- **Customers** — customer information and city
- **Orders** — order transactions and quantities
- **Products** — product information and prices

The tables are combined using CustomerID and ProductID to create an analysis-ready dataset.

```python
import pandas as pd

customers = pd.DataFrame({
    'CustomerID': [1, 2, 3, 4],
    'CustomerName': ['Alice', 'Bob', 'Charlie', 'David'],
    'City': ['London', 'Manchester', 'Leeds', 'Bristol']
})

products = pd.DataFrame({
    'ProductID': [101, 102, 103],
    'ProductName': ['Laptop', 'Mouse', 'Keyboard'],
    'Price': [900, 25, 60]
})

orders = pd.DataFrame({
    'OrderID': [1, 2, 3, 4],
    'CustomerID': [1, 1, 2, 3],
    'ProductID': [101, 102, 103, 102],
    'Quantity': [1, 2, 1, 3]
})
```

## Data Quality Checks
Before analysing the data, I checked the dataset structure, missing values, duplicate records, and basic data validity.

```python
# Check dataset structure
print('Customers:', customers.shape)
print('Orders:', orders.shape)
print('Products:', products.shape)

# Check missing values
print(customers.isna().sum())
print(orders.isna().sum())
print(products.isna().sum())

# Check duplicates
print('Duplicate customers:', customers.duplicated().sum())
print('Duplicate orders:', orders.duplicated().sum())
print('Duplicate products:', products.duplicated().sum())
```

## Data Preparation

The Orders table was merged with Customers and Products to combine transaction, customer, and product information.

Sales was calculated as:

**Sales = Quantity × Price**

```python
df = orders.merge(customers, on = 'CustomerID', how = 'left')
df = df.merge(products, on = 'ProductID', how = 'left')

#create sale column
df['Sales'] = df['Quantity'] * df['Price']
df
```

## Product Analysis
This analysis identifies which products contribute most to sales and which products have the highest unit volume.

```python
product_analysis = (
    df.groupby('ProductName')
      .agg(
          TotalQuantity = ('Quantity', 'sum'),
          TotalSales = ('Sales', 'sum')
      )
      .reset_index()
)

product_analysis.sort_values(
    'TotalSales',
    ascending = False
)
```

### Finding
Laptop generated the highest revenue at £900, representing approximately 83% of total revenue in this dataset.

## Customer Analysis
Customer-level analysis was performed to identify high-value customers based on total spending, order volume, and quantity purchased.

```python
customer_analysis = (
    df.groupby('CustomerName')
    .agg(
        TotalOrders = ('OrderID', 'count'),
        TotalQuantity = ('Quantity', 'sum'),
        TotalSales = ('Sales', 'sum')
    )
    .reset_index()
)

customer_analysis.sort_values(
    'TotalSales',
    ascending = False
)
```

### Finding
Alice was the highest-value customer by total spending, generating £950 and contributing approximately 88% of total revenue.

## City Analysis
City-level analysis was used to identify markets with the highest sales revenue and sales volume.

```python
city_analysis = (
    df.groupby('City')
    .agg(
        TotalQuantity = ('Quantity', 'sum'),
        TotalSales = ('Sales', 'sum')
    )
    .reset_index()
)

city_analysis.sort_values(
    'TotalSales',
    ascending = False
)
```

### Finding
London generated the highest sales revenue at £950, making it the strongest-performing market in the current dataset.

## Customer Revenue Contribution
Revenue share was calculated to understand how total sales are distributed across customers.

```python
customer_analysis['RevenueShare'] = (
    customer_analysis['TotalSales'] / customer_analysis['TotalSales'].sum() * 100
)

customer_analysis.sort_values(
    'RevenueShare',
    ascending = False
)
```

### Finding
Alice contributed approximately 88% of total revenue in the current dataset, indicating a high concentration of revenue among a single customer.

## Key Insights
1. **Product:** Laptop generated £900 in revenue, representing approximately 83% of total revenue and making it the main revenue contributor in this dataset.
2. **Customer:** Alice was the highest-value customer by total spending, generating £950 and contributing approximately 88% of total revenue.
3. **Market:** London generated £950 in sales revenue, making it the strongest-performing market in the current dataset.
